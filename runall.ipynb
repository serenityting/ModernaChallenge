```python
import os
import sys
import subprocess
from pathlib import Path
from IPython.display import Image, display
from qiskit_algorithms import QAOA
```

```python
cwd = Path.cwd()

if (cwd / "src").exists():
    PROJECT_ROOT = cwd
    SRC = cwd / "src"
elif cwd.name == "src":
    PROJECT_ROOT = cwd.parent
    SRC = cwd
else:
    raise RuntimeError("Cannot find src directory.")

print("Project Root:", PROJECT_ROOT)
print("Source Folder:", SRC)
```

```python
import subprocess
import sys
from pathlib import Path

def run_script(script, args=None):
    if args is None:
        args = []

    script_path = SRC / script

    print(f"\n{'='*80}")
    print(f"Running {script}")
    print(f"{'='*80}")

    result = subprocess.run(
        [sys.executable, str(script_path)] + args,
        cwd=PROJECT_ROOT,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    print(result.stdout)

    if result.returncode != 0:
        print(f"\n❌ {script} failed with exit code {result.returncode}")
        return False

    print(f"✅ {script} completed")
    return True
```

```python
scripts = [
    "qubo_builder.py",
    "solvers.py",
    "metrics.py",
    "benchmark.py",
    "noise_robustness.py",
    "scaling_analysis.py",
    "make_plots.py",
]
```
